### dependent & independent features

```python
X = df.drop("FWI", axis = 1)
y = df["FWI"]
```

```python
X.head()
```

```python
y.head()
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.25, random_state = 15)
```

```python
X_train.corr()
```

### redundancy, multicollinearity, overfitting

```python
print(X_train.corr().iloc[0, 2])
```

```python
def correlation_for_dropping(df, threshold):
    columns_to_drop = set()
    corr = df.corr()
    for i in range(len(corr.columns)):
        for j in range(i):
            if abs(corr.iloc[i, j]) > threshold:
                columns_to_drop.add(corr.columns[i])
    return columns_to_drop
```

```python
columns_dropping = correlation_for_dropping(X_train, 0.85)
```

```python
columns_dropping
```

```python
X_train.drop(columns_dropping, axis = 1, inplace = True)
X_test.drop(columns_dropping, axis = 1, inplace = True)
```

```python
X_train.shape
```

```python
X_test.shape
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
```

```python
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
plt.subplots(figsize = (20, 5))
plt.subplot(1, 2, 1)
sns.boxplot(data = X_train)
plt.title("X_train")
plt.subplot(1, 2, 2)
sns.boxplot(data = X_train_scaled)
plt.title("X_train_scaled")
plt.show()
```

### Linear

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
linear = LinearRegression()
linear.fit(X_train_scaled, y_train)
y_pred = linear.predict(X_test_scaled)
mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
score = r2_score(y_test, y_pred)
print("mae: ", mae)
print("mse: ", mse)
print("score: ", score)
plt.scatter(y_test, y_pred)
plt.show()
```

### Lasso

```python
from sklearn.linear_model import Lasso
lasso = Lasso()
lasso.fit(X_train_scaled, y_train)
y_pred = lasso.predict(X_test_scaled)
mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
score = r2_score(y_test, y_pred)
print("mae: ", mae)
print("mse: ", mse)
print("score: ", score)
plt.scatter(y_test, y_pred)
plt.show()
```

### Ridge

```python
from sklearn.linear_model import Ridge
ridge = Ridge()
ridge.fit(X_train_scaled, y_train)
y_pred = ridge.predict(X_test_scaled)
mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
score = r2_score(y_test, y_pred)
print("mae: ", mae)
print("mse: ", mse)
print("score: ", score)
plt.scatter(y_test, y_pred)
plt.show()
```

### ElasticNet

```python
from sklearn.linear_model import ElasticNet
elsnet = ElasticNet()
elsnet.fit(X_train_scaled, y_train)
y_pred = elsnet.predict(X_test_scaled)
mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
score = r2_score(y_test, y_pred)
print("mae: ", mae)
print("mse: ", mse)
print("score: ", score)
plt.scatter(y_test, y_pred)
plt.show()
```

### Lasso Cross Validation

```python
from sklearn.linear_model import LassoCV
```

```python
lassocv = LassoCV(cv = 5)
lassocv.fit(X_train_scaled, y_train)
y_pred = lassocv.predict(X_test_scaled)
mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
score = r2_score(y_test, y_pred)
print("mae: ", mae)
print("mse: ", mse)
print("score: ", score)
plt.scatter(y_test, y_pred)
plt.show()
```
